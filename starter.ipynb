```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from time import perf_counter
```

```python
# Implement baseline algorithm
def baseline_Max_subarray(arr):
    max_so_far = float('-inf')
    n = len(arr)
    for i in range(n):
        for j in range(i, n):
            current_sum = 0
            for k in range(i, j + 1):
                current_sum += arr[k]
            if current_sum > max_so_far:
                max_so_far = current_sum
    return max_so_far

import numpy
import time as time
arr = numpy.random.randint(-100, 100, size=1000)

start = time.perf_counter()
baseline_Max_subarray(arr)
end = time.perf_counter()

elapsed = end - start

# Collect target execution times for different input sizes
records = []

for k in range(100, len(arr) + 1, 100):
    start = time.perf_counter()
    baseline_Max_subarray(arr[:k])
    end = time.perf_counter()
    records.append({"k": k, "time": end - start})

df_base = pd.DataFrame(records)

sns.lineplot(data=df_base, x="k", y="time")
plt.xlabel("k")
plt.ylabel("time (s)")
plt.title("Execution Time of sumFirstK Algorithm")
plt.show()
```

```python
# Implement target algorithm
def kadaneMaxsubarray(arr):
    max_so_far = float('-inf')
    current_max = 0
    for i in range(len(arr)):
        current_max = current_max + arr[i]
        if current_max < 0:
            current_max = 0
        if current_max > max_so_far:
            max_so_far = current_max
    return max_so_far

import numpy
import time as time
arr = numpy.random.randint(-100, 100, size=1000)

start = time.perf_counter()
kadaneMaxsubarray(arr)
end = time.perf_counter()

elapsed = end - start

# Collect target execution times for different input sizes
records = []

for k in range(100, len(arr) + 1, 100):
    start = time.perf_counter()
    kadaneMaxsubarray(arr[:k])
    end = time.perf_counter()
    records.append({"k": k, "time": end - start})

df_base = pd.DataFrame(records)

sns.lineplot(data=df_base, x="k", y="time")
plt.xlabel("k")
plt.ylabel("time (s)")
plt.title("Execution Time of sumFirstK Algorithm")
plt.show()
```
